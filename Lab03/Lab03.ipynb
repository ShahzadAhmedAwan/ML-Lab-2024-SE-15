## Task 3 · Clean the data: the six types of problem

Fix the problems **in this order** (text first, then duplicates):

- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3a: missing values
print("T3a: missing values per column (before)")
print(df.isna().sum())

# ads with no brand or no model are useless, so drop them
rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
n_missing_removed = rows_before - len(df)
print("T3a: rows removed (no Brand or no Model) =", n_missing_removed)

# text columns get "Unknown", Year and KMs Driven stay NaN
df = df.fillna({"Registered City": "Unknown", "Fuel": "Unknown", "Transaction Type": "Unknown"})
print("T3a: rows left =", len(df))

print("T3a: missing values per column (after)")
print(df.isna().sum())
```

```python
# T3b: formatting
cities_raw = df["Registered City"].nunique()
print("T3b: city spellings before =", cities_raw)

# spaces and capital letters in city names
df["Registered City"] = df["Registered City"].str.strip().str.title()
print("T3b: city spellings after strip + title =", df["Registered City"].nunique())

# remove "Rs" and commas from Price (regex=False because they are plain text)
n_rs = df["Price"].str.contains("Rs").sum()
print("T3b: prices written with 'Rs' =", n_rs)
df["Price"] = df["Price"].str.replace("Rs", "", regex=False).str.replace(",", "", regex=False).str.strip()
print("T3b: prices still with 'Rs' or a comma =", df["Price"].str.contains("Rs|,").sum())
```

```python
# T3c: data types
price_dtype_before, year_dtype_before = df["Price"].dtype, df["Year"].dtype

df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")    # capital I so it can hold NaN

print("T3c: Price dtype =", df["Price"].dtype, "(was", str(price_dtype_before) + ")")
print("T3c: Year dtype =", df["Year"].dtype, "(was", str(year_dtype_before) + ")")

# back to T2h: the same comparison that failed on text now works on numbers
print("T3c: ads with Price > 1,000,000 =", (df["Price"] > 1_000_000).sum())
```

```python
# T3d: inconsistent categories
print("T3d: different cities before =", df["Registered City"].nunique())
print("T3d: 'Sukkar' ads =", (df["Registered City"] == "Sukkar").sum())
print("T3d: 'Corrolla Altis' ads =", (df["Model"] == "Corrolla Altis").sum())

n_sukkar = (df["Registered City"] == "Sukkar").sum()
n_corrolla = (df["Model"] == "Corrolla Altis").sum()
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
print("T3d: different cities after =", df["Registered City"].nunique())
```

```python
# T3e: duplicates (after the text is cleaned, so spelling cannot hide them)
rows_before_dup = len(df)
n_dups = df.duplicated().sum()
print("T3e: exact duplicate ads =", n_dups)

df = df.drop_duplicates()    # assign back, it returns a new table
print("T3e: rows after dropping duplicates =", len(df))
```

```python
# T3f: invalid values
# impossible kilometres become NaN, the rest of the ad is still useful
kms = df["KMs Driven"]
bad_kms = (kms < 100) | (kms > 1_000_000) | kms.isin([12345, 123456])

kms_mean_before = kms.mean()
df.loc[bad_kms, "KMs Driven"] = np.nan
kms_mean_after = df["KMs Driven"].mean()

print("T3f: invalid KMs Driven set to NaN =", bad_kms.sum())
print("T3f: mean KMs Driven before =", round(kms_mean_before), "| after =", round(kms_mean_after))

# fake prices and very expensive Suzuki/Daihatsu ads are dropped
fake_price = df["Price"].isin([12_345_678, 87_654_321])
too_expensive = df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
drop_rows = fake_price | too_expensive

print("T3f: fake prices =", fake_price.sum())
print("T3f: Suzuki/Daihatsu above 5,000,000 =", too_expensive.sum())
print("T3f: rows dropped =", drop_rows.sum())

df = df[~drop_rows].reset_index(drop=True)
```

```python
# T3g: self-check
print("T3g: final shape =", df.shape)
print("T3g: matches (20484, 9)? ->", df.shape == (20484, 9))
assert df.shape == (20484, 9), "self-check failed"

# one line per type of dirty data, every number is computed in the cells above
cleaning_log = pd.DataFrame({
    "Type": ["1 Missing values", "2 Formatting", "3 Data types", "4 Inconsistent categories", "5 Duplicates", "6 Invalid values"],
    "Found": [f"{n_missing_removed} ads without Brand or Model",
              f"{n_rs} 'Rs' prices (after T3a), {cities_raw} city spellings",
              f"Price was {price_dtype_before}, Year was {year_dtype_before}",
              f"{n_sukkar} 'Sukkar' + {n_corrolla} 'Corrolla Altis' ads",
              f"{n_dups} exact duplicate ads",
              f"{bad_kms.sum()} impossible KMs, {drop_rows.sum()} fake/too-high prices"],
    "Fix": ["dropna(subset), fillna('Unknown')",
            "str.strip().str.title(), str.replace",
            "astype('int64') and astype('Int64')",
            "replace({...})",
            "drop_duplicates()",
            "KMs -> NaN, bad-price rows dropped"]})
print("T3g: cleaning log (fixed in this order)")
print(cleaning_log.to_string(index=False))
```

```python
# T3g (extra check): duplicates that only appear after Type 6 (the manual puts duplicates before invalid values)
n_dups_after = df.duplicated().sum()
print("T3g: duplicate rows left after Type 6 =", n_dups_after)
print("T3g: shape kept at", df.shape, "because the self-check requires (20484, 9); rows if they were dropped =", len(df) - n_dups_after)
```

**Cleaning notes**

- **Order used:** missing → formatting → types → categories → duplicates → invalid values. Text was cleaned before the duplicate check, so `KARACHI` and `Karachi` ads could not hide from `drop_duplicates()`.
- **Missing values:** the 2,448 ads without `Brand` or `Model` were dropped. After that no column had any NaN left, so the missing `Year` and `KMs Driven` values were in those same ads.
- **Invalid vs outlier:** impossible values (fake prices, impossible kilometres) were fixed or dropped by a rule; unusual but possible prices (Task 4g) were only flagged, because the lab says to check outliers, not to delete them automatically.
- **Invalid kilometres** were set to NaN and not deleted, because I do not know the real value and the rest of the ad is still useful.
- **Duplicates that appear later:** 128 ads became identical only after Type 6 set their impossible kilometres to NaN (they differed only in that wrong value). The manual fixes the order (duplicates before invalid values) and the self-check shape (20484, 9), so I kept them. The extra check above counts them, and they could be dropped in Lab 04 if needed.

## Task 4 · Analyse

- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.

- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?

```python
# T4a: price in lakh
df["Price_lakh"] = df["Price"] / 100_000
print("T4a: Price and Price_lakh, first 5")
display(df[["Price", "Price_lakh"]].head())

# T4b: age of the car (the lab uses 2020 as "today")
df["Age"] = 2020 - df["Year"]
print("T4b: Year and Age, first 5")
display(df[["Year", "Age"]].head())
```

```python
# T4c: most common brands
print("T4c: 5 most common brands")
print(df["Brand"].value_counts().head(5))

# T4d: the 5 cities with the most ads, exactly as the task is worded
print("T4d: 5 cities with the most ads")
print(df["Registered City"].value_counts().head(5))

# "Unknown" is a placeholder I created in Task 3a and not a real city, so I also show real cities only
known_city = df["Registered City"] != "Unknown"
print("T4d (known cities only): 5 cities with the most ads")
print(df.loc[known_city, "Registered City"].value_counts().head(5))
print("T4d: ads with city 'Unknown' =", (~known_city).sum())

# T4e: median price per fuel type
print("T4e: median Price_lakh per fuel type, highest first")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))
```

```python
# T4f: Corolla GLI, model years 2010 to 2018
gli = df[(df["Model"] == "Corolla GLI") & df["Year"].between(2010, 2018)]

gli_by_year = gli.groupby("Year")["Price_lakh"].agg(ads="count", median_price="median")
gli_by_year["change_vs_prev_year"] = gli_by_year["median_price"].diff()

print("T4f: Corolla GLI, median price (lakh) by year")
display(gli_by_year)

# average loss per year of age, from 2018 down to 2010
price_2018 = gli_by_year.loc[2018, "median_price"]
price_2010 = gli_by_year.loc[2010, "median_price"]
loss_per_year = (price_2018 - price_2010) / (2018 - 2010)
print("T4f: average loss per year of age (lakh) =", round(loss_per_year, 2))
```

**T4f answer.** The median Corolla GLI falls from 23.2 lakh (2018) to 12.25 lakh (2010), so it loses about **1.37 lakh for every extra year of age**. The loss is not constant: the steps from 2010 to 2013 are about 0.8 to 1.1 lakh, while the last two steps (2016 to 2018) are 2.0 and 2.2 lakh, so a car loses the most rupees while it is new. Each year has 64 to 110 ads, so the medians are fairly reliable. This compares different model years at one point in time, it does not follow one car over time.

```python
# T4g: 5 most expensive ads
top5 = df.nlargest(5, "Price")[["Brand", "Model", "Year", "KMs Driven", "Condition", "Price_lakh"]].copy()

# compare each ad with the median price of its own model
model_median = df.groupby("Model")["Price_lakh"].transform("median")
top5["times_model_median"] = (top5["Price_lakh"] / model_median[top5.index]).round(1)

# a model covers many years, so also compare with the same model AND the same year
by_model_year = df.groupby(["Model", "Year"])["Price_lakh"]
top5["times_model_year_median"] = (top5["Price_lakh"] / by_model_year.transform("median")[top5.index]).round(1)
top5["ads_same_model_year"] = by_model_year.transform("size")[top5.index]

print("T4g: 5 most expensive ads")
display(top5)

# the comparison figures quoted in the answer below
prius_2015 = df[(df["Model"] == "Prius") & (df["Year"] == 2015)]["Price_lakh"].sort_values(ascending=False)
surf_2007 = df[(df["Model"] == "Surf") & (df["Year"] == 2007)]["Price_lakh"].sort_values(ascending=False)
lc_2005 = df[(df["Model"] == "Land Cruiser") & (df["Year"] == 2005)]["Price_lakh"].sort_values(ascending=False)
print("T4g: 2015 Prius, highest 2 prices (lakh) =", prius_2015.head(2).tolist())
print("T4g: 2007 Surf, highest 2 prices (lakh) =", surf_2007.head(2).tolist())
print("T4g: 2005 Land Cruiser, all prices (lakh) =", lc_2005.tolist())
```

**T4g answer.** Yes, some of them look wrong. All five are Toyotas, but a model-wide median is only a rough yardstick, because a model such as the Land Cruiser covers many model years. So I also compared each ad with the median of the same model and the same year.

- The **2015 Prius at 357.5 lakh** and the **2007 Surf at 410 lakh** are almost certainly mistakes: both ask about 20 times their model median, and 14.3 and 7.7 times the median of their own model year. The next most expensive 2015 Prius asks 32.5 lakh and the next 2007 Surf 53.5 lakh.
- The **2005 Land Cruiser at 520 lakh** is suspicious too: it asks 9 times the median of its model year (the other two 2005 ads are 58 and 43 lakh), and it costs more than the 2015 one (420 lakh) although it is ten years older and `Used`.
- The **2017 Land Cruiser at 670 lakh** and the **2015 one at 420 lakh** are outliers, not necessarily errors: they are only 2.3 and 1.8 times the median of their model year, which is high but possible. Those medians come from only 3 ads each, so they are rough. These two should be checked, not deleted automatically.

I did not use the `Condition` label as evidence, because it is unreliable in this data: 2,035 of the 3,889 `New` ads have a model year before 2015 (see bonus B3).

I flagged these ads and did not delete them: Task 3f only asks to drop the listed fake prices, and the self-check shape must stay (20484, 9).

```python
# T4h: Cash vs Installment/Leasing
by_transaction = df.groupby("Transaction Type").agg(
    ads=("Price", "size"),
    median_price_lakh=("Price_lakh", "median"),
    median_age=("Age", "median"),
    median_kms=("KMs Driven", "median"))

print("T4h: Cash vs Installment/Leasing")
display(by_transaction)

# same comparison for young cars only, so age cannot explain the gap
young = df[df["Age"] <= 6]
print("T4h: median Price_lakh, cars aged 6 years or less")
print(young.groupby("Transaction Type")["Price_lakh"].median())

# same model, other transaction type: what does an installment ad ask compared with cash ads?
cash_median = df.loc[df["Transaction Type"] == "Cash"].groupby("Model")["Price"].median()
installment = df[df["Transaction Type"] == "Installment/Leasing"]
price_ratio = installment["Price"] / installment["Model"].map(cash_median)
print("T4h: installment price / median cash price of the same model (25th, 50th and 75th percentile; the 50th is the median ratio)")
print(price_ratio.quantile([0.25, 0.5, 0.75]).round(2))
```

**T4h answer.** Cash has 19,634 ads with a median of 6.9 lakh, but Installment/Leasing has only 850 ads with a median of 3.0 lakh, although those cars are much newer (median age 4 vs 13 years). The same model asks only about 0.34 times the cash price (median ratio), so an installment price is probably a down payment or one installment and not the full price (the data cannot confirm this), which means the two groups should not be compared directly.
