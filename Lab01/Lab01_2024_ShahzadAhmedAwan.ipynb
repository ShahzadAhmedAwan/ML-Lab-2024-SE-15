## Task 6: Tuple vs list

- **Question:** What is the main difference between a tuple and a list?
- **Answer:** A **list is mutable**: it can be changed after it is created. A **tuple is immutable**: once created, it cannot be changed. Trying to change a tuple raises a `TypeError`.

<table style="margin-left:0; margin-right:auto; border-collapse:collapse; text-align:left;">
<tr><th style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Feature</th><th style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">List</th><th style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Tuple</th></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Syntax</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;"><code>[1, 2, 3]</code></td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;"><code>(1, 2, 3)</code></td></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Change an item</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Allowed</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Not allowed (<code>TypeError</code>)</td></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Add or remove items</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;"><code>append</code>, <code>remove</code>, <code>pop</code></td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Not possible</td></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Best for</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Data that changes (a list of marks)</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Data that must stay fixed (image size, coordinates)</td></tr>

- **What they share:** both are ordered, can hold mixed types, and support indexing and slicing.
- **Why it matters in ML:** a tuple protects values such as an image's `(height, width)` from accidental changes.
- **Demo below:** the list changes successfully, and the tuple raises an error.

```python
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

my_list[0] = 99                      # works: lists can be changed
print("Q6 list:", my_list)

try:
    my_tuple[0] = 99                 # fails: tuples cannot be changed
except TypeError as error:
    print("Q6 tuple error:", error)
```

## Task 7: Get the domain from an email

- **Question:** Write a function that grabs the domain from `"user@domain.com"`.
- **Approach:** `"user@domain.com".split("@")` returns `['user', 'domain.com']`. Index `[1]` is the domain.
- **Expected result:** `domain.com`

```python
def get_domain(email):
    """Return the part of an email address after the '@' sign."""
    # split("@") gives [username, domain], so [1] is the domain
    return email.split("@")[1]

print("Q7:", get_domain("user@domain.com"))
```

## Task 8: Is `dog` in a string?

- **Question:** Write a function that returns `True` if the word `dog` is in a string (case-insensitive).
- **Approach:** Convert the text to lower case with `.lower()`, then use the `in` test. This way `Dog` and `DOG` also match.
- **Expected result:** `True` when "dog" is present in any case, otherwise `False`.

```python
def find_dog(text):
    """Return True if 'dog' appears in text, ignoring upper/lower case."""
    # lower() makes the check case-insensitive
    return "dog" in text.lower()

print("Q8:", find_dog("Is there a dog here?"))
print("Q8 (uppercase):", find_dog("A DOG is friendly"))
print("Q8 (no dog):", find_dog("The cat is sleeping"))
```

## Task 9: Count `dog` in a string

- **Question:** Write a function that counts how many times `dog` occurs in a string (case-insensitive).
- **Approach:** Convert the text to lower case first, then use `.count("dog")`. Without `.lower()`, `Dog` and `DOG` would be missed.
- **Expected result:** `2` for `"This dog runs faster than the other dog dude!"`

```python
def count_dog(text):
    """Return how many times 'dog' occurs in text, ignoring case."""
    # lower() first so Dog and DOG are counted too
    return text.lower().count("dog")

print("Q9:", count_dog("This dog runs faster than the other dog dude!"))
print("Q9 (mixed case):", count_dog("Dog, dog and DOG"))
```

## Task 10: Caught speeding

- **Question:** Write `caught_speeding(speed, is_birthday)` that returns `No Ticket` (60 or less), `Small Ticket` (61 to 80) or `Big Ticket` (81 or more). On your birthday, all limits are 5 mph higher.

<table style="margin-left:0; margin-right:auto; border-collapse:collapse; text-align:left;">
<tr><th style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Result</th><th style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">No birthday</th><th style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Birthday</th></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">No Ticket</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">60 or less</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">65 or less</td></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Small Ticket</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">61 to 80</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">66 to 85</td></tr>
<tr><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">Big Ticket</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">81 or more</td><td style="text-align:left; vertical-align:top; padding:6px 14px; border:1px solid #9a9a9a;">86 or more</td></tr>
</table>

- **Approach:** Set an `allowance` of 5 on a birthday (otherwise 0) and add it to both limits, so the logic is written only once. A loop then runs the test cases, including the edge cases.

```python
def caught_speeding(speed, is_birthday):
    """Return 'No Ticket', 'Small Ticket' or 'Big Ticket'.
    On a birthday, each speed limit is 5 mph higher."""
    # extra speed allowed on a birthday
    allowance = 5 if is_birthday else 0

    if speed <= 60 + allowance:
        return "No Ticket"
    elif speed <= 80 + allowance:
        return "Small Ticket"
    else:
        return "Big Ticket"

# (speed, is_birthday) test cases, including the boundaries
tests = [(60, False), (61, False), (80, False), (81, False),
         (65, True), (66, True), (85, True), (86, True)]

# run every test case and print the result
for speed, birthday in tests:
    print("Q10: speed =", speed, "| birthday =", birthday, "->", caught_speeding(speed, birthday))
```

### Bonus 1: Count `dog` with punctuation

- **Question:** Make `count_dog` also count `dog,` and `dog!` (punctuation).
- **Approach:** Split the text into words, strip punctuation from each word, then compare it with `"dog"`. Only whole words match, so `hotdog` is not counted.
- **Expected result:** `4` for `"Dog, dog! hotdog dog. DOG"`

```python
def count_dog_punctuation(text):
    """Count the word 'dog' (any case), ignoring punctuation around it."""
    count = 0
    # check each word on its own, so 'hotdog' is not counted
    for word in text.lower().split():
        # strip() removes punctuation from both ends of the word
        if word.strip(".,!?") == "dog":
            count += 1
    return count

print("Bonus 1:", count_dog_punctuation("Dog, dog! hotdog dog. DOG"))
```

### Bonus 2: Palindrome

- **Question:** Write `is_palindrome(word)` using slicing `word[::-1]`.
- **Approach:** A palindrome reads the same forwards and backwards. `word[::-1]` reverses a string, so compare it with the original.
- **Expected result:** `True` for `radar` and `Level`, `False` for `python`.

```python
def is_palindrome(word):
    """Return True if word reads the same forwards and backwards (ignoring case)."""
    word = word.lower()
    # word[::-1] is the reversed string
    return word == word[::-1]

print("Bonus 2:", is_palindrome("radar"), is_palindrome("Level"), is_palindrome("python"))
```

- **Question:** Store 5 students as a list of dicts and print the topper's name.
- **Approach:** Each student is a dictionary with a `name` and `marks`. Loop through the list and keep the student with the highest marks.

```python
students = [
    {"name": "Aisha", "marks": 88},
    {"name": "Bilal", "marks": 95},
    {"name": "Hina",  "marks": 91},
    {"name": "Omar",  "marks": 84},
    {"name": "Sara",  "marks": 97},
]

# start with the first student, replace whenever someone scores higher
topper = students[0]
for student in students:
    if student["marks"] > topper["marks"]:
        topper = student

print("Bonus 3:", topper["name"], "with", topper["marks"], "marks")
```
